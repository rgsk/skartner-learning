```python
from fractions import Fraction
import math
import numpy as np
from typing import Dict, Any


def geom_sum(a, r, n):
    return (a * (1 - r**n)) / (1 - r)


def decimal_to_fraction(x: float, max_denominator: int = 10**6) -> Fraction:
    """
    Convert a decimal number to a Fraction in lowest terms.

    Parameters:
      x               – the decimal number (float)
      max_denominator – the largest denominator to search (default 1_000_000)

    Returns:
      A fractions.Fraction instance in simplest form.
    """
    return Fraction(x).limit_denominator(max_denominator)


def dist_two_points(x1, y1, x2, y2):
    return math.sqrt((x2 - x1) ** 2 + (y2 - y1) ** 2)


def C(n, r):
    if r == 0:
        return 1
    if r > n // 2:
        return C(n, n - r)
    return n * C(n - 1, r - 1) // r


def bernoulli_trial(n, r, p=None, p_not=None):
    if p is None and p_not is None:
        raise Exception("Either p or p_not should be defined")
    if p is None:
        p = 1 - p_not
    if p_not is None:
        p_not = 1 - p
    return C(n, r) * p**r * (p_not) ** (n - r)


def binomial_distribution(n, start, end=None, p=None, p_not=None):
    if end == None:
        end = n
    ans = 0
    for i in range(start, end + 1):
        ans += bernoulli_trial(n, i, p, p_not)
    return ans


def geometric_pmf(p, x):
    return (1 - p) ** (x - 1) * p


def geometric_distribution(p, start, end):
    result = 0
    for x in range(start, end + 1):
        result += geometric_pmf(p, x)
    return result


def arcsec(x: float) -> float:
    """Return arcsec(x) in radians. Domain: |x| >= 1."""
    if abs(x) < 1:
        raise ValueError("arcsec domain error: |x| must be ≥ 1")
    return math.acos(1 / x)


def arccsc(x: float) -> float:
    """Return arccsc(x) in radians. Domain: |x| >= 1."""
    if abs(x) < 1:
        raise ValueError("arccsc domain error: |x| must be ≥ 1")
    return math.asin(1 / x)


def arccot(x: float) -> float:
    """Return arccot(x) in radians. Domain: |x| >= 1."""
    return math.atan(1 / x)


def avg_rate_of_change(f, x1, x2):
    return (f(x2) - f(x1)) / (x2 - x1)


def law_of_cosines(b, c, A):
    return math.sqrt(b**2 + c**2 - 2 * b * c * math.cos(math.radians(A)))


def law_of_cosines_angle(a, b, c):
    return math.degrees(math.acos((b**2 + c**2 - a**2) / (2 * b * c)))


def area_of_general_triangle(a, b, C):
    return (a * b * math.sin(math.radians(C))) / 2


def proj_x_a(x, a):
    # a is a vector
    return (np.dot(x, a) / np.dot(a, a)) * a


def dist(a, b):
    return np.linalg.norm(a - b)


def proj_x_A(x, A):
    # A is a basis of a span
    return A @ np.linalg.inv(A.T @ A) @ A.T @ x


def poisson_pmf(lambda_value, x):
    """
    Calculate the Poisson PMF.

    Parameters:
    - lambda_value (float): The expected number of occurrences (λ).
    - x (int): The actual number of occurrences.

    Returns:
    - float: The probability of observing x occurrences.
    """
    return (lambda_value**x) * math.exp(-lambda_value) / math.factorial(x)


def poisson_range(lambda_value, start, end):
    result = 0
    for x in range(start, end + 1):
        result += poisson_pmf(lambda_value, x)
    return result


def negative_binomial(p, x, r):
    return C(x - 1, r - 1) * (1 - p) ** (x - r) * p**r


def negative_binomial_distribution(p, r, start, end):
    result = 0
    for x in range(start, end + 1):
        result += negative_binomial(p=p, x=x, r=r)
    return result


def expected_value(x_values, fx_values):
    if len(x_values) != len(fx_values):
        raise Exception("length must be same")
    n = len(x_values)
    result = 0
    for i in range(n):
        result += x_values[i] * fx_values[i]
    return result


def counter(values):
    mp: Dict[Any, int] = {}
    for v in values:
        mp[v] = mp.get(v, 0) + 1
    return mp


def fact(x: int):
    return math.factorial(x)


def trinomial(n, p, x):
    arrangements = fact(n) / (fact(x[0]) * fact(x[1] * fact(x[2])))
    return arrangements * p[0] ** x[0] * p[1] ** x[1] * p[2] ** x[2]


def multinomial(n, p, x):
    coefficient = fact(n) / math.prod(fact(xi) for xi in x)
    probability = math.prod(pi**xi for pi, xi in zip(p, x))
    return coefficient * probability
```
